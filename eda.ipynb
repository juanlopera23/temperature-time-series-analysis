### Moving Average Analysis Conclusion

The daily temperature series shows frequent short-term fluctuations throughout the year, which makes the overall pattern more difficult to observe directly.

Using a **7-day moving average** reduces much of this daily variability and makes the underlying seasonal pattern clearer. Temperatures generally increase from winter toward summer, reach their highest levels around July and August, and then decrease again toward the end of the year.

The moving average therefore provides a clearer representation of the short-term temperature trend while preserving the general behavior of the original time series.

### January Outlier Analysis Conclusion

The **Z-score method** did not identify any strong outliers in January, since no observation exceeded **±3 standard deviations** from the monthly mean.

However, the **IQR method** identified **four potential outliers**: three unusually high temperatures and one unusually low temperature.

The boxplot confirms these IQR-based outliers visually, showing the observations outside the lower and upper fences.

This difference shows that outlier detection depends on the method used. In this case, the IQR method is more sensitive to extreme January temperatures than the Z-score method.

```python
df_monthly_avg=pd.DataFrame(df.groupby(["Year","Month"])["AvgTemperature"].mean().round(2))
```

```python
df_monthly_avg.head(24)
```

```python
index_min= df_monthly_avg["AvgTemperature"].idxmin()
temperature_min=df_monthly_avg["AvgTemperature"].min()

index_max= df_monthly_avg["AvgTemperature"].idxmax()
temperature_max=df_monthly_avg["AvgTemperature"].max()

print(f"The month-year with the lowest average temperature: {index_min[0]}/{index_min[1]}")
print(f"the temperature was: {temperature_min}")
print("-"*60)
print(f"The month-year with the highest average temperature: {index_max[0]}/{index_max[1]}")
print(f"the temperature was: {temperature_max}")
```

### Monthly Analysis Conclusion

Between 1995 and 2019, the monthly average temperatures in Columbus show a clear seasonal pattern, with lower temperatures during winter and higher temperatures during summer.

The lowest monthly average temperature was recorded in **February 2015, at 19.36 °F**, while the highest monthly average temperature occurred in **July 2012, at 80.25 °F**.

These results show a significant variation in temperature throughout the year.

### Daily Temperature Extremes Conclusion

Between 1995 and 2019, the coldest day recorded in Columbus was **January 16, 2009**, with an average temperature of **-3.8 °F**.

The hottest day was **July 7, 2012**, with an average temperature of **89.6 °F**.

These results show the wide range of daily temperatures experienced in Columbus over the analyzed period.
